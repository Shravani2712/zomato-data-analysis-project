```python
import pandas as pd
import folium
from IPython.display import display, HTML, IFrame, clear_output
import matplotlib.pyplot as plt
import ipywidgets as widgets
```

```python
# ====== Dashboard Title ======
display(HTML("<h1 style='color:green; text-align:center'>Restaurant Mini Dashboard</h1>"))

# ====== Interactive Widgets ======
city_options = ['All'] + sorted(df['city'].dropna().unique().tolist())
city_dropdown = widgets.Dropdown(options=city_options, description='City:')
rating_slider = widgets.FloatRangeSlider(
    value=[0, 5],
    min=0,
    max=5,
    step=0.1,
    description='Rating:',
    continuous_update=False,
    readout_format='.1f'
)

# Container for filters
filter_box = widgets.HBox([city_dropdown, rating_slider])
display(filter_box)

# ====== Function to Update Dashboard ======
def update_dashboard(city, rating_range):
    clear_output(wait=True)  # clear previous outputs
    display(HTML("<h1 style='color:green; text-align:center'>Restaurant Mini Dashboard</h1>"))
    display(filter_box)
    
    # ====== Filter Data ======
    filtered_df = df.copy()
    if city != 'All':
        filtered_df = filtered_df[filtered_df['city'] == city]
    filtered_df = filtered_df[(filtered_df['rating'] >= rating_range[0]) & (filtered_df['rating'] <= rating_range[1])]
    
    # ====== Summary Stats ======
    total_restaurants = len(filtered_df)
    avg_rating = filtered_df['rating'].mean() if len(filtered_df) > 0 else 0
    top_city = filtered_df['city'].mode()[0] if len(filtered_df) > 0 else 'N/A'
    
    summary_html = f"""
    <div style='border:1px solid gray; padding:10px; width:400px; margin-bottom:10px;'>
        <h3>Summary</h3>
        <p><b>Total Restaurants:</b> {total_restaurants}</p>
        <p><b>Average Rating:</b> {avg_rating:.2f}</p>
        <p><b>Top City in Filter:</b> {top_city}</p>
    </div>
    """
    display(HTML(summary_html))
    
    # ====== Charts ======
    plt.close('all')
    
    # Pie chart - Restaurants per city
    city_counts = filtered_df['city'].value_counts()
    if not city_counts.empty:
        plt.figure(figsize=(5,5))
        city_counts.plot.pie(autopct='%1.1f%%', startangle=90,
                             colors=['lightblue','lightgreen','lightcoral','khaki','lightgrey'])
        plt.title('Restaurants Distribution by City')
        plt.ylabel('')
        plt.show()
    
    # Bar chart - Average rating per city
    avg_rating_city = filtered_df.groupby('city')['rating'].mean().sort_values(ascending=False)
    if not avg_rating_city.empty:
        plt.figure(figsize=(6,4))
        avg_rating_city.plot(kind='bar', color='skyblue')
        plt.ylabel('Average Rating')
        plt.title('Average Rating per City')
        plt.xticks(rotation=45)
        plt.show()
    
    # Column chart - Total votes per city
    votes_city = filtered_df.groupby('city')['votes'].sum().sort_values(ascending=False)
    if not votes_city.empty:
        plt.figure(figsize=(6,4))
        votes_city.plot(kind='bar', color='salmon')
        plt.ylabel('Total Votes')
        plt.title('Total Votes per City')
        plt.xticks(rotation=45)
        plt.show()
    
    # ====== Map ======
    def get_marker_color(rating):
        try:
            rating = float(rating)
            if rating >= 4.0:
                return 'green'
            elif rating >= 2.5:
                return 'orange'
            else:
                return 'red'
        except:
            return 'blue'
    
    if not filtered_df.empty:
        m = folium.Map(location=[filtered_df['latitude'].mean(), filtered_df['longitude'].mean()], zoom_start=12)
        for i, row in filtered_df.iterrows():
            name = row['restaurant_name'] if pd.notna(row['restaurant_name']) else 'Unknown'
            cuisine = row['cuisines'] if pd.notna(row['cuisines']) else 'Not specified'
            rating = row['rating'] if pd.notna(row['rating']) else 'No rating'
            price_range = row['price_range'] if pd.notna(row['price_range']) else 'N/A'
            votes = row['votes'] if pd.notna(row['votes']) else 'N/A'
            city_name = row['city'] if pd.notna(row['city']) else 'Unknown'
            
            if pd.notna(row['latitude']) and pd.notna(row['longitude']):
                folium.Marker(
                    location=[row['latitude'], row['longitude']],
                    popup=(
                        f"<b>{name}</b> ({cuisine})<br>"
                        f"Rating: {rating} ★<br>"
                        f"Price Range: {price_range}<br>"
                        f"Votes: {votes}<br>"
                        f"City: {city_name}"
                    ),
                    icon=folium.Icon(color=get_marker_color(rating), icon='cutlery', prefix='fa')
                ).add_to(m)
        
        # Save and display map
        m.save('filtered_map.html')
        display(IFrame('filtered_map.html', width=800, height=500))
    else:
        display(HTML("<p>No restaurants match the selected filters.</p>"))

# ====== Connect Widgets ======
interactive_dashboard = widgets.interactive(update_dashboard, city=city_dropdown, rating_range=rating_slider)
```
