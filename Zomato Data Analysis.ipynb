```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
stacked_data = pd.crosstab(df['city'], df['has_online_delivery']).head(10)

stacked_data.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)

plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.title("Restaurants by City & Online Delivery")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
grouped_data = pd.crosstab(
    df['has_table_booking'],
    df['has_online_delivery']
)

grouped_data.plot(
    kind='bar',
    figsize=(6,4)
)

plt.xlabel("Table Booking")
plt.ylabel("Restaurant Count")
plt.title("Table Booking vs Online Delivery")
plt.show()
```

```python
year_growth = df['year_opening'].value_counts().sort_index()

plt.figure(figsize=(8,4))
plt.plot(year_growth.index, year_growth.values, marker='o')

plt.xlabel("Year")
plt.ylabel("Restaurants Opened")
plt.title("Restaurant Growth Over Years")
plt.grid(True)
plt.show()
```

```python
delivery_counts = df['has_online_delivery'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    delivery_counts,
    labels=delivery_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Online Delivery Distribution")
plt.show()
```

```python
df['rating_bucket'] = pd.cut(
    df['rating'],
    bins=[0,2,3,4,5],
    labels=['Poor','Average','Good','Excellent']
)

rating_counts = df['rating_bucket'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    rating_counts,
    labels=rating_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Restaurant Rating Distribution")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12,8))
# Bar: Top cities
top_cities = df['city'].value_counts().head(5)
sns.barplot(
    x=top_cities.index,
    y=top_cities.values,
    ax=axes[0,0]
)
axes[0,0].set_title("Top 5 Cities")

# Line: Growth
axes[0,1].plot(year_growth.index, year_growth.values, marker='o')
axes[0,1].set_title("Growth Over Years")

# Pie: Delivery
axes[1,0].pie(delivery_counts, labels=delivery_counts.index, autopct='%1.1f%%')
axes[1,0].set_title("Online Delivery")

# Bar: Rating bucket
sns.countplot(
    x='rating_bucket',
    data=df,
    ax=axes[1,1]
)
axes[1,1].set_title("Ratings Distribution")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import folium
```

```python
def get_marker_color(rating):
    try:
        rating = float(rating)
        if rating >= 4.0:
            return 'green'
        elif rating >= 2.5:
            return 'orange'
        else:
            return 'red'
    except:
        return 'blue'  # fallback if rating is missing or invalid

# Create base map centered at mean coordinates
m = folium.Map(location=[df['latitude'].mean(), df['longitude'].mean()], zoom_start=12)

# Add restaurant markers
for i, row in df.iterrows():
    # Handle missing data
    name = row['restaurant_name'] if pd.notna(row['restaurant_name']) else 'Unknown'
    cuisine = row['cuisines'] if pd.notna(row['cuisines']) else 'Not specified'
    rating = row['rating'] if pd.notna(row['rating']) else 'No rating'
    price_range = row['price_range'] if pd.notna(row['price_range']) else 'N/A'
    votes = row['votes'] if pd.notna(row['votes']) else 'N/A'
    city = row['city'] if pd.notna(row['city']) else 'Unknown'
    
    if pd.notna(row['latitude']) and pd.notna(row['longitude']):
        folium.Marker(
            location=[row['latitude'], row['longitude']],
            popup=(
                f"<b>{name}</b> ({cuisine})<br>"
                f"Rating: {rating} ★<br>"
                f"Price Range: {price_range}<br>"
                f"Votes: {votes}<br>"
                f"City: {city}"
            ),
            icon=folium.Icon(color=get_marker_color(rating), icon='cutlery', prefix='fa')
        ).add_to(m)

# Display map in Jupyter Notebook
m
```
